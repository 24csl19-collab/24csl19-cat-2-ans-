```python

import pandas as pd
import zipfile
import os
from google.colab import files

uploaded = files.upload()

filename = next(iter(uploaded))

if filename.endswith(".zip"):
    with zipfile.ZipFile(filename, "r") as z:
        csv_files = [f for f in z.namelist()
                     if f.lower().endswith(".csv")]
        if not csv_files:
            raise ValueError("No CSV file found inside ZIP")
        with z.open(csv_files[0]) as f:
            df = pd.read_csv(f)
else:
    df = pd.read_csv(filename)

print("Dataset Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

display(df.head())
print("\nMissing Values:")
print(df.isnull().sum())
```

```python

import numpy as np
import re

# Convert the rating column to numeric
df["rating"] = pd.to_numeric(df["rating"], errors="coerce")

# Remove rows without a valid target rating
df = df.dropna(subset=["rating"]).copy()

# Keep valid rating values
df = df[df["rating"].between(1, 5)].copy()

# Convert price and percentage columns to numeric
def to_number(series):
    return pd.to_numeric(
        series.astype(str).str.replace(
            r"[₹,% ,]", "", regex=True
        ),
        errors="coerce"
    )

df["discounted_price"] = to_number(df["discounted_price"])
df["actual_price"] = to_number(df["actual_price"])
df["discount_percentage"] = to_number(
    df["discount_percentage"]
)
df["rating_count"] = to_number(df["rating_count"])

# Clean review text and remove URLs
def clean_text(text):
    text = str(text)
    text = re.sub(r"https?://\S+|www\.\S+", " ", text)
    text = re.sub(r"\s+", " ", text)
    return text.strip()

text_columns = [
    "product_name", "category",
    "about_product", "review_title", "review_content"
]

for col in text_columns:
    df[col] = df[col].fillna("").apply(clean_text)

# Combine product information and review text
df["combined_text"] = df[text_columns].agg(" ".join, axis=1)

NUM_COLS = [
    "discounted_price",
    "actual_price",
    "discount_percentage",
    "rating_count"
]

# Fill missing numerical features with their medians
for col in NUM_COLS:
    df[col] = df[col].fillna(df[col].median())

print("Cleaned dataset shape:", df.shape)
print("Remaining missing values:")
print(df[NUM_COLS + ["rating"]].isnull().sum())

print("\nRating distribution:")
print(df["rating"].describe())
```

```python

from sklearn.model_selection import GroupShuffleSplit
from sklearn.preprocessing import StandardScaler

# Inputs and target
X_text = df["combined_text"].astype(str)
X_num = df[NUM_COLS].astype("float32")
y = df["rating"].astype("float32")

# Group split by product ID
splitter = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

train_idx, test_idx = next(
    splitter.split(X_text, y, groups=df["product_id"])
)

X_train_text = X_text.iloc[train_idx]
X_test_text = X_text.iloc[test_idx]

X_train_num = X_num.iloc[train_idx]
X_test_num = X_num.iloc[test_idx]

y_train = y.iloc[train_idx]
y_test = y.iloc[test_idx]

# Scale numerical features using training data only
scaler = StandardScaler()

X_train_num = scaler.fit_transform(X_train_num)
X_test_num = scaler.transform(X_test_num)

# Tokenize text into numerical sequences
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

vectorizer = layers.TextVectorization(
    max_tokens=10000,
    output_mode="int",
    output_sequence_length=120,
    standardize="lower_and_strip_punctuation"
)

vectorizer.adapt(
    tf.data.Dataset.from_tensor_slices(
        X_train_text.to_numpy()
    ).batch(32)
)

print("Vocabulary size:", vectorizer.vocabulary_size())
print("Example token sequence:")
print(vectorizer(X_train_text.iloc[:1]).numpy())
print("Training records:", len(train_idx))
print("Testing records:", len(test_idx))
```

```python

text_input = keras.Input(
    shape=(),
    dtype=tf.string,
    name="review_text"
)

numeric_input = keras.Input(
    shape=(len(NUM_COLS),),
    dtype=tf.float32,
    name="numeric_features"
)

# Text processing
x = vectorizer(text_input)

x = layers.Embedding(
    input_dim=vectorizer.vocabulary_size(),
    output_dim=32,
    mask_zero=True
)(x)

x = layers.GlobalAveragePooling1D()(x)
x = layers.Dense(32, activation="relu")(x)
x = layers.Dropout(0.30)(x)

# Combine text and numerical features
combined = layers.Concatenate()([x, numeric_input])

combined = layers.Dense(32, activation="relu")(combined)
combined = layers.Dropout(0.20)(combined)
combined = layers.Dense(16, activation="relu")(combined)

# Predict a continuous rating
output = layers.Dense(1, activation="linear")(combined)

model = keras.Model(
    inputs=[text_input, numeric_input],
    outputs=output
)

model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.001),
    loss="mse",
    metrics=["mae"]
)

model.summary()
```

```python

early_stop = keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=6,
    restore_best_weights=True
)

history = model.fit(
    {
        "review_text": X_train_text.to_numpy(),
        "numeric_features": X_train_num
    },
    y_train.to_numpy(),
    validation_split=0.15,
    epochs=50,
    batch_size=16,
    callbacks=[early_stop],
    verbose=1
)

print("Training completed!")
```

```python

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

predictions = model.predict(
    {
        "review_text": X_test_text.to_numpy(),
        "numeric_features": X_test_num
    },
    verbose=0
).flatten()

# Rating scale is 1 to 5
predictions = np.clip(predictions, 1, 5)

mae = mean_absolute_error(y_test, predictions)
rmse = np.sqrt(mean_squared_error(y_test, predictions))
r2 = r2_score(y_test, predictions)

print("MODEL EVALUATION")
print("----------------")
print(f"MAE  : {mae:.4f}")
print(f"RMSE : {rmse:.4f}")
print(f"R2   : {r2:.4f}")

results = pd.DataFrame({
    "Actual Rating": y_test.to_numpy(),
    "Predicted Rating": np.round(predictions, 2)
})

display(results.head(10))
```

```python

def predict_product_rating(
    product_name,
    category,
    review_title,
    review_content,
    about_product="",
    discounted_price=499,
    actual_price=699,
    discount_percentage=29,
    rating_count=100
):
    # Prepare unseen text
    text = " ".join([
        clean_text(product_name),
        clean_text(category),
        clean_text(about_product),
        clean_text(review_title),
        clean_text(review_content)
    ])

    # Prepare numeric features in the same column order
    numeric = pd.DataFrame([[
        discounted_price,
        actual_price,
        discount_percentage,
        rating_count
    ]], columns=NUM_COLS)

    numeric = scaler.transform(numeric)

    # Predict - explicitly cast input to object/float32 to avoid dtype resolution errors
    pred = model.predict(
        {
            "review_text": np.array([text], dtype=object),
            "numeric_features": numeric.astype("float32")
        },
        verbose=0
    )[0, 0]

    pred = float(np.clip(pred, 1, 5))

    print("Product:", product_name)
    print("Review:", review_content)
    print(f"Predicted Rating: {pred:.2f} / 5 stars")

    return pred


# Example unseen review
predict_product_rating(
    product_name="Wireless Bluetooth Headphones",
    category="Electronics",
    review_title="Good sound quality",
    review_content="The sound is clear and the battery lasts long.",
    about_product="Wireless headphones with rechargeable battery",
    discounted_price=1499,
    actual_price=1999,
    discount_percentage=25,
    rating_count=150
)
```

```python

import numpy as np
from sklearn.metrics import accuracy_score, classification_report

# Actual test ratings
actual = np.asarray(y_test).round().astype(int)

# Get model predictions
predicted = model.predict(
    {
        "review_text": X_test_text.to_numpy(),
        "numeric_features": X_test_num
    },
    verbose=0
).flatten()

# Convert predictions to star ratings
predicted_stars = np.clip(
    np.floor(predicted + 0.5), 1, 5
).astype(int)

# Calculate genuine exact-star accuracy
accuracy = accuracy_score(actual, predicted_stars) * 100

print(f"Exact Star Accuracy: {accuracy:.2f}%")
print("\nClassification Report:")
print(classification_report(
    actual, predicted_stars, zero_division=0
))
```
